```python
downloader.download("uzbek plov rice dish carrots and meat on plate",
                    limit=100,
                    output_dir="/content/pilaf_images",
                    adult_filter_off=False,
                    force_replace=False,
                    timeout=60
                    )
```

```python
from google.colab import drive
dataset_path = "/content/drive/MyDrive/food101_dataset"
```

```python
import os, shutil

source_dir ='/content/pilaf_images/uzbek plov rice dish carrots and meat on plate'
target_dir = '/content/drive/MyDrive/food101_dataset/pilaf'

os.makedirs(target_dir, exist_ok=True)

for filename in os.listdir(source_dir):
  src_file = os.path.join(source_dir, filename)
  dst_file = os.path.join(target_dir, filename)
  shutil.move(src_file, dst_file)
```
